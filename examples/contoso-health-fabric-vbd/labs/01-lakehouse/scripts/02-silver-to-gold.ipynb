# Lab 01 · Silver → Gold

> Verified against Microsoft Learn on 2026-09-15.

**Objective:** aggregate silver claims to a monthly-per-provider gold table for Warehouse + PBI consumption.

````python
from pyspark.sql import functions as F

df = spark.table('silver_claims')
gold = (df
  .withColumn('month', F.date_trunc('month', 'claim_date'))
  .groupBy('provider_id', 'month')
  .agg(
     F.sum('claim_amount').alias('total_paid'),
     F.count('*').alias('claim_count'),
     F.avg('claim_amount').alias('avg_claim'))
)
gold.write.mode('overwrite').format('delta').saveAsTable('gold_claims_by_provider_month')
````

## Checkpoint
```sql
SELECT * FROM gold_claims_by_provider_month ORDER BY total_paid DESC LIMIT 10;
```
